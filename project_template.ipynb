### To what extent will the introduction of mandatory helmet requirement change the usage patterns towards helmet use?

The introduction of the helmet mandate in 2023 for Dutch light moped (snorfiets) is utilized to make a forecast of change in usage patterns towards helmet use. The research towards attitude and the modal change effects could bring valuable insides for a new mandate for e-bikes.

#### Effect of a helmet mandate on underage e-bike use and road incidents

A helmet mandate would raise helmet use very likely. However, it may also push many underage riders off the e-bike. The net effect on fatal and serious incidents is probably positive (Schmidt et al., 2026). The displaced trips move to slower modes rather than to faster vehicles since they have no legal alternative within the age range 12-17 years old.

#### Helmet usage pattern

After the Dutch light moped (snorfiets) mandate of 1 January 2023, roadside counts found 99% helmet use (Schmidt et al., 2026). Among the riders who kept riding the light moped, those wearing a helmet "always" rose from 9% to 67% (Schmidt et al., 2026). Voluntary use among young cyclists is very low: 77.8% of 12-24-year-olds never wear a helmet, and only 0.4% do so on an e-bike (TeamAlert, 2020). Among e-bike and fatbike casualties at emergency departments, helmet use was 6% and 3% (VeiligheidNL, 2024).

#### Attitudes of underage riders

Young riders value the e-bike for speed and low effort. One rider named not having to wear a helmet as a reason to prefer it over a scooter (TeamAlert, 2023). Among 12-24-year-olds, 46.9% say others' opinions matter if they wear a helmet and 48.3% fear remarks (TeamAlert, 2020). In focus groups with 13-18-year-olds, nobody wore a helmet, some said they would stop riding an e-bike if one were required, and one said she would not wear it if she were the only one (TeamAlert, 2023). After the light moped mandate, 20% of surveyed riders stopped among those who continued, daily or weekly use fell from about three-quarters to just over half, 34% rode less and 15% later stopped (Schmidt et al., 2026). Attitudes improved among continuing riders: helmet use seen as a habit rose from 19.7% to 50.9%, and the perception that others wear helmets rose from 18.2% to 63.6% (Schmidt et al., 2026).



#### Where trips go

Replaced light moped trips went mainly to cars (over 50%), e-bikes (33%), regular bicycles (20%) and walking (20%) fatbikes and speed pedelecs were chosen by fewer than 5% (Schmidt et al., 2026). The combined moped and light moped fleet fell by about 6%, with light mopeds down 24% and mopeds up 29% (Schmidt et al., 2026). Some young e-bike riders already tune their bikes to 32-37 km/h (TeamAlert, 2023), This means a move to faster or tuned vehicles is plausible, but the sources do not quantify it.




#### Forecast for fatal and serious incidents

- **Direct effect:**

  Helmets cut serious head injury risk by about 60% and fatal head trauma by 70% (TeamAlert, 2020). Among the seriously injured light moped riders, the part with severe head injury fell from 55% to 30% (Schmidt et al., 2026). Brain injury accounted for 22% of e-bike and 17% of fatbike casualties (VeiligheidNL, 2024).
  
- **Exposure effect:**

  After the 2023 mandate, light moped crashes fell 38%, rider deaths 38% and severe injuries 50% (Schmidt et al., 2026).
- **Net effect:**

  SWOV projected that full bicycle helmet compliance would save about 85 lives and 2,500ÔÇô2,600 serious injuries per year (SWOV, 2022) judge the net effect of the light moped mandate likely favourable, but some casualties may now occur on other modes and the size of that shift is unknown. A shift to faster vehicles would reduce the gain.


### Prediction Of Helmet Adaption After Mandate For Underaged Users
#### Objective
based on this qualitive analysis a quantative objective is structured as follow:

This analysis predicts how helmet use is likely to spread among 12- to 17-year-old e-bike and fatbike riders once helmets become mandatory. Qualitative data on youth attitudes toward helmets were used to set upper limits on adoption under three scenarios. The resulting projections provide the basis for estimating how many severe and fatal head injuries the mandate could prevent.

#### Approach

The projections used an asymptotic growth model using parametres based of the evaluation of the 2023 Dutch light moped (snorfiets) mandate and surveys of youth attitudes. In each scenario, adoption starts from the current voluntary helmet use of 0.4% and rises month by month with an assumption addaption rate of k = 0.4 toward a scenario-specific ceiling:

* **Best case:** Adoption follows the pattern observed after the light moped mandate, where 67% of continuing riders reported "always" wearing a helmet.
* **Expected case:** The best-case ceiling is reduced to account for social pressure. In the surveys, 48.3% of young people feared comments from others about wearing a helmet, and 46.9% valued others' opinions on the matter. Applying a 48% reduction to the 67% ceiling yields a maximum compliance of 34.8%.
* **Worst case:** Underage riders lack a legal transport alternative. After the light moped mandate, cars absorbed more than half of the displaced trips, an option unavailable to this age group. Widespread avoidance behaviour was therefore assumed, whith an estimated caps compliance at 15%. 

#### Results

Projected compliance differs substantially across scenarios see figure TODO. At months 1, 6 and 12, compliance is roughly 22%, 61% and 67% in the best case; 12%, 32% and 35% in the expected case; and 5%, 14% and 15% in the worst case.

The adoption rate is assumed (k = 0.4) in all scenarios, each reaches about 90% of its ceiling by month six. The ceiling therefore drives the differences: by month 12, the best and worst cases are about 52 percentage points apart.

The expected-case ceiling treats the 48% of youth who fear peer comments as a proportional reduction of the best case. This is a huge simplification, since not everyone who fears comments will skip the helmet. The worst-case ceiling of 15% is an assumption reflecting that 12-17 year-olds, unlike adult moped riders, have no legal alternative for their trips, which makes avoidance more likely.

The results are scenario bounds and can't be seen as point forecasts. Limitations include self-reported helmet use, extrapolation from moped riders to younger users, and the absence of enforcement effects. These bounds provide the input for estimating prevented severe and fatal head injuries.
