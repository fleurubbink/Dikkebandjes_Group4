# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

What impact will the introduction of the compulsory helmet rule for underage e-bike users have on road safety?
    To what extent will the number of accidents involving e-bike users change?
    To what extent will the introduction of a compulsory helmet rule affect the road safety of other road users?
    To what extent will underage e-bike users comply with this new legislation?
        To what extent will this new legislation have a positive/negative impact on the behaviour of underage e-bike users?
        To what extent will increased enforcement affect the behaviour of underage e-bike users?
    To what extent will the introduction of compulsory helmet use reduce/increase the use of e-bikes?
